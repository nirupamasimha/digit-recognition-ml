```python
import os
import numpy as np
import keras
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Input, Dense, Dropout, Flatten, Conv2D, MaxPooling2D, BatchNormalization
#from keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
```

```python
# Load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Reshape and Normalize target images to 0-1 float bounds
x_train = x_train.reshape(x_train.shape[0], 28, 28, 1).astype("float32") / 255.0
x_test = x_test.reshape(x_test.shape[0], 28, 28, 1).astype("float32") / 255.0

# Convert label vectors into binary matrix formats
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)

print(f"Data processing complete. Training shape context: {x_train.shape}")
```

```python
# Initialize Model Architecture matching Keras 3 operational criteria
model = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, kernel_size=(3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2, 2)),
    
    Conv2D(64, kernel_size=(3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2, 2)),
    
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.4),
    Dense(10, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# Train Model directly using stable tensor batches
history = model.fit(x_train, y_train, batch_size=128, epochs=15, validation_data=(x_test, y_test))
```

```python
# Save the modern weights native file
model.save('digit_cnn_model.keras')
print("Model trained and stored successfully under 'digit_cnn_model.keras'!")
```
